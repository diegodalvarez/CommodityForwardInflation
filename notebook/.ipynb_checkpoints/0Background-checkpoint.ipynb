# Replicating St. Louis Federal Reserve Graphs

In the writeup *Oil prices and expected inflation* [here](https://fredblog.stlouisfed.org/2020/04/oil-prices-and-expected-inflation/?utm_source=series_page&utm_medium=related_content&utm_term=related_resources&utm_campaign=fredblog) from the Federal Reserve of St. Louis the authors model the relationship in the other direction comparing the raw spot prices for forward inflation. They only use WTI and 5y5y forward but it can be extended to other spot energy prices. 

```python
path   = os.path.join(data_path, "FRED", "CombinedData.parquet")
df_raw = pd.read_parquet(path = path, engine = "pyarrow")
```

```python
df_inf = (df_raw
    .loc[lambda x: x.group == "inflation"]
    .drop(columns = ["group", "ticker"])
    .rename(columns = {"value": "forward_inf"}))
```

```python
df_combined = (df_raw
    .loc[lambda x: x.group != "inflation"]
    .drop(columns = ["group"])
    .merge(right = df_inf, how = "inner", on = ["date"]))
```

```python
tickers   = df_combined.ticker.drop_duplicates().sort_values().to_list()
fig, axes = plt.subplots(ncols = 3, nrows = 2, figsize = (20,8))

for ticker, ax in zip(tickers, axes.flatten()):

    (df_combined
        .loc[lambda x: x.ticker == ticker]
        .plot(
            alpha  = 0.6,
            ax     = ax,
            kind   = "scatter", 
            y      = "forward_inf", 
            x      = "value",
            xlabel = "Spot Price ($)",
            ylabel = "5y5y Forward Inflation (%)",
            title  = RENAMER[ticker]))

    ax.yaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x:.1f}%"))
    ax.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f"${x:,.0f}"))

fig.suptitle("Comparing Spot Price vs. 5y5y Forward Inflation")
plt.tight_layout()
```

In this case the FRED graphs use spot prices and breakeven inflation, comparing yield to spot prices is semi-stable since both prices are subject to some sort of range-boundedness. Although the comparison needs to be rigously examined through regression. In this case take daily differences in spot prices and normalize by their 10d exponentially weighted standard deviation, for the 5y5y forward yield take the log differences. 

```python
ols_path = os.path.join(data_path, "FactorData", "SpotLinearModels.pkl")
log_path = os.path.join(data_path, "FactorData", "SpotLogitModels.pkl")

with open(ols_path, "rb") as f: ols_models = pickle.load(f)
with open(log_path, "rb") as f: log_models = pickle.load(f)
```

```python
df_ols_params = FactorModel().get_param(ols_models, model_type = "OLS")
df_log_params = FactorModel().get_param(log_models, model_type = "Logit")
```

Below is a table of the OLS regression

```python
display(df_ols_params
    .assign(
        ticker = lambda x: x.name.str.split("_").str[0],
        lag    = lambda x: x.name.str.split("_").str[1])
    .loc[lambda x: x.param_name != "const"]
    .drop(columns = ["name", "r2", "param_name"])
    .melt(id_vars = ["ticker", "lag"])
    .replace(RENAMER)
    .rename(columns = {
        "ticker"  : "Energy",
        "lag"     : "Lag",
        "variable": ""})
    .pivot(index = "Energy", columns = ["", "Lag"], values = "value"))
```

As well as logistic regression 

```python
display(df_log_params
    .assign(
        ticker = lambda x: x.name.str.split("_").str[0],
        lag    = lambda x: x.name.str.split("_").str[1])
    .loc[lambda x: x.param_name != "const"]
    .drop(columns = ["name", "param_name", "pseudo_r2"])
    .melt(id_vars = ["ticker", "lag"])
    .replace(RENAMER)
    .rename(columns = {
        "ticker"  : "Energy",
        "lag"     : "Lag",
        "variable": ""})
    .pivot(index = "Energy", columns = ["", "Lag"], values = "value"))
```

Under the No Lag (instantanoues comparison they don't)

```python
fig, axes = plt.subplots(figsize = (9,5))

(df_combined
    .set_index("date")
    .dropna()
    .groupby("ticker")
    .apply(lambda x: x.value.rolling(window = 252).corr(x.forward_inf))
    .to_frame(name = "correlation")
    .reset_index()
    .rename(columns = {"ticker": ""})
    .pivot(index = "date", columns = "", values = "correlation")
    .rename(columns = RENAMER)
    .plot(
        xlabel = "",
        ylabel = "Correlation (%)",
        ax     = axes,
        title  = "Rolling 252 day correlation between Spot Energy Prices vs. 5y5y forward inflation"))

axes.yaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x:.0%}"))
```

```python
df_diff_corr = (df_combined
    .set_index("date")
    .dropna()
    .groupby("ticker")
    .apply(lambda x: x.diff())
    .reset_index()
    .set_index("date")
    .groupby("ticker")
    .apply(lambda x: x.value.rolling(window = 252).corr(x.forward_inf))
    .to_frame(name = "corr_val"))
```

```python
fig, axes = plt.subplots(figsize = (9,5))


(df_diff_corr
    .reset_index()
    .rename(columns = {"ticker": ""})
    .pivot(index = "date", columns = "", values = "corr_val")
    .rename(columns = RENAMER)
    .plot(
        xlabel = "",
        ylabel = "Correlation (%)",
        ax     = axes,
        title  = "Rolling 252 day correlation between Spot Energy vs. 5y5y forward inflation (differenced)"))

axes.yaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x:.0%}"))
plt.show()
```

```python
df_coverage = (df_diff_corr
    .reset_index()
    .dropna()
    .assign(corr_group = lambda x: np.where(x.corr_val > 0, "positive", "negative"))
    .drop(columns = ["corr_val"])
    .groupby(["ticker", "corr_group"])
    .agg("count")
    .reset_index()
    .pivot(index = "ticker", columns = "corr_group", values = "date")
    .assign(
        cuml  = lambda x: x.sum(axis = 1),
        ratio = lambda x: x.positive / x.cuml)
    [["ratio"]])
```

```python
ticker_map = (df_ticker
    .set_index("ticker")
    .name
    .to_dict())
```

```python
display(df_diff_corr
    .reset_index()
    .drop(columns = ["date"])
    .dropna()
    .groupby("ticker")
    ["corr_val"]
    .agg(["min", "max", "mean", "median"])
    .merge(right = df_coverage, how = "inner", on = ["ticker"])
    .rename(columns = {
        "min"   : "Lowest",
        "max"   : "Highest",
        "mean"  : "Average",
        "median": "Median",
        "ratio" : "% > 0"})
    .reset_index()
    .assign(name = lambda x: x.ticker.map(ticker_map))
    .drop(columns = ["ticker"])
    .rename(columns = {"name": ""})
    .set_index(""))
```
